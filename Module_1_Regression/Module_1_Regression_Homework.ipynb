# 📈 Ödev 1: Lineer ve Lojistik Regresyon
Bu ödevde, derste öğrendiğiniz Makine Öğrenmesi temellerini kullanarak gerçek hayat senaryolarını kodlayacaksınız. 
- Önce bir evin büyüklüğüne göre **Fiyat Tahmini** yapan bir Lineer Regresyon kuracağız.
- Daha sonra, yaşı ve geliri verilen bir müşterinin şirketi terk edip etmeyeceğini tahmin eden bir **Lojistik Regresyon** (Sınıflandırma) kuracağız.

> ⚠️ Aşağıdaki `init` (başlangıç) hücresini mutlaka çalıştırın.

```python
# init
def check_hw1(model_linear, mse_degeri, model_logistic, y_pred_log):
    score = 0
    print("--- Ödev 1 Test Sonuçları ---")
    if model_linear is not None and hasattr(model_linear, 'coef_'): score+=25; print("✅ Görev 1: Lineer model başarıyla eğitildi.")
    else: print("❌ Görev 1 Hatalı.")
    if mse_degeri is not None and mse_degeri < 50: score+=25; print("✅ Görev 2: MSE formülü ve hesabı doğru.")
    else: print("❌ Görev 2 Hatalı.")
    if model_logistic is not None and hasattr(model_logistic, 'coef_'): score+=25; print("✅ Görev 3: Lojistik model eğitildi.")
    else: print("❌ Görev 3 Hatalı.")
    if y_pred_log is not None and len(y_pred_log)>0: score+=25; print("✅ Görev 4: Lojistik model ile tahmin başarıyla yapıldı.")
    else: print("❌ Görev 4 Hatalı.")
    print(f"\nToplam Puanınız: {score}/100")
```
