Apply the ML basics from class. First predict house prices from floor area with linear regression; then classify whether a customer will leave using age and income. Run init first.

```python

def _evaluate_hw1(model_linear, mse_degeri, model_logistic, y_pred_log, context=None):
    import numpy as np
    from sklearn.linear_model import LinearRegression, LogisticRegression
    from sklearn.metrics import mean_squared_error
    results = []
    X_train = np.array([[50], [60], [80], [100], [120]])
    y_train = np.array([150, 180, 240, 300, 360])
    X_test = np.array([[70], [90], [110]])
    y_test = np.array([210, 270, 330])
    try:
        pred = np.asarray(model_linear.predict(X_test))
        ok_linear = isinstance(model_linear, LinearRegression) and pred.shape == y_test.shape and np.allclose(pred, y_test, rtol=0, atol=1e-6) and np.allclose(model_linear.predict(X_train), y_train, rtol=0, atol=1e-6)
    except Exception: ok_linear = False
    results.append((25, bool(ok_linear), "Görev 1: eğitilmiş lineer model / Task 1: fitted linear model"))
    try:
        ok_mse = ok_linear and np.asarray(mse_degeri).ndim == 0 and not isinstance(mse_degeri, (bool, np.bool_)) and np.isfinite(mse_degeri) and mse_degeri >= 0 and np.isclose(mse_degeri, mean_squared_error(y_test, pred), rtol=1e-6, atol=1e-8)
    except Exception: ok_mse = False
    results.append((25, bool(ok_mse), "Görev 2: tahminlerden doğru MSE / Task 2: correct MSE from predictions"))
    X_train_log = np.array([[22, 30], [25, 40], [45, 100], [50, 120]])
    y_train_log = np.array([1, 1, 0, 0])
    X_test_log = np.array([[24, 35], [48, 110]])
    try:
        expected_pred = np.asarray(model_logistic.predict(X_test_log))
        ok_log = isinstance(model_logistic, LogisticRegression) and np.array_equal(model_logistic.predict(X_train_log), y_train_log) and expected_pred.shape == (2,)
    except Exception: ok_log = False
    results.append((25, bool(ok_log), "Görev 3: eğitilmiş lojistik model / Task 3: fitted logistic model"))
    try:
        actual = np.asarray(y_pred_log)
        ok_pred = ok_log and actual.shape == (2,) and np.isin(actual, [0, 1]).all() and np.array_equal(actual, expected_pred)
    except Exception: ok_pred = False
    results.append((25, bool(ok_pred), "Görev 3 tahmin: iki müşterinin sınıfı / Task 3 predictions: classes of both customers"))
    return results

def check_hw1(model_linear, mse_degeri, model_logistic, y_pred_log):
    return _show_results(_evaluate_hw1(model_linear, mse_degeri, model_logistic, y_pred_log))

def _show_results(results):
    score = 0
    print("--- Ödev Test Sonuçları / Homework Test Results ---")
    for points, ok, label in results:
        print(("✅ Başarılı / Passed: " if ok else "❌ Hatalı veya eksik / Incorrect or missing: ") + label)
        if ok: score += points
    print(f"Toplam Puan / Total Score: {score}/100")
    return score
```
