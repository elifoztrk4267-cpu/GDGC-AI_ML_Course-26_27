Yapay zeka modellerinin başarısı, onlara verilen verinin temizliğine bağlıdır! Bu ödevde meşhur [Spaceship Titanic](https://www.kaggle.com/competitions/spaceship-titanic) veri setinin çok minyatür bir versiyonu üzerinde çalışarak veriyi modele nasıl hazırlayacağımızı göreceğiz.

> ⚠️ Aşağıdaki `init` hücresini çalıştırın.

```python
# init
def check_hw3(df):
    score = 0
    print("--- Ödev 3 Test Sonuçları ---")
    if df is not None:
        if df['Age'].isnull().sum() == 0: score+=35; print("✅ Görev 1.1: Yaş verisindeki boşluklar dolduruldu!")
        else: print("❌ Görev 1.1 Hatalı. 'Age' içinde hala boşluklar var.")
        
        if df['VIP'].isnull().sum() == 0: score+=35; print("✅ Görev 1.2: VIP verisindeki boşluklar dolduruldu!")
        else: print("❌ Görev 1.2 Hatalı.")
        
        if 'PassengerId' not in df.columns: score+=30; print("✅ Görev 2: Gereksiz sütun (PassengerId) silindi!")
        else: print("❌ Görev 2 Hatalı. Sütun hala duruyor.")
    else:
        print("❌ Dataframe (`df_temiz`) bulunamadı.")
    print(f"\nToplam Puanınız: {score}/100")
```
