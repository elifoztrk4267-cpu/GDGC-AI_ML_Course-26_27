# 🚀 Ödev 3: Spaceship Titanic Preprocess
Eksik verileri doldurup DataFrame'i temizleyeceğiz.

```python
# init
def check_hw3(df):
    score = 0
    print("--- Ödev 3 Test Sonuçları ---")
    if df is not None:
        if df['Age'].isnull().sum() == 0: score+=35; print("✅ Yaş verileri doldurulmuş!")
        else: print("❌ Yaş verileri eksik.")
        if df['VIP'].isnull().sum() == 0: score+=35; print("✅ VIP verileri doldurulmuş!")
        else: print("❌ VIP verileri eksik.")
        if 'PassengerId' not in df.columns: score+=30; print("✅ PassengerId silinmiş!")
        else: print("❌ PassengerId silinmemiş.")
    else: print("❌ Dataframe bulunamadı.")
    print(f"Toplam Puan: {score}/100")
```

## Görev: Eksik Verileri Doldurma ve Sütun Silme
Eksik `Age` ve `VIP` verilerini doldurun. `PassengerId` sütununu silin. Sonucu `df_temiz` değişkenine atayın.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'PassengerId': ['01', '02', '03'], 'Age': [24, np.nan, 30], 'VIP': [False, True, np.nan]})

# KODUNUZU BURAYA YAZIN
df_temiz = None
```
