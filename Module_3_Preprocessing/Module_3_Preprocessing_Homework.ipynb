```python
_HW3_EXPECTED_ROWS = 5

def _evaluate_hw3(df, context=None):
    import numpy as np
    import pandas as pd
    context = {} if context is None else context
    rows = context.get("_HW3_EXPECTED_ROWS", 3)
    if rows == 5:
        age = np.array([24., 33., 30., 45., 33.]); vip = [False, True, False, False, False]
    else:
        age = np.array([24., 27., 30.]); vip = [False, True, False]
    structure = isinstance(df, pd.DataFrame) and len(df) == len(age) and df.index.equals(pd.RangeIndex(len(age))) and not df.columns.duplicated().any()
    try:
        actual_age = df["Age"].to_numpy()
        ok_age = structure and pd.api.types.is_numeric_dtype(df["Age"]) and np.isfinite(actual_age).all() and np.allclose(actual_age, age, rtol=0, atol=1e-8)
    except Exception: ok_age = False
    try:
        values = df["VIP"].tolist()
        ok_vip = structure and all(isinstance(v, (bool, np.bool_)) for v in values) and values == vip
    except Exception: ok_vip = False
    ok_drop = structure and set(df.columns) == {"Age", "VIP"}
    return [(35, bool(ok_age), "Görev 1.1: Age ortalama ile doldurulmuş, satırlar korunmuş / Task 1.1: Age mean-imputed, rows preserved"), (35, bool(ok_vip), "Görev 1.2: VIP False ile doldurulmuş, mevcut değerler korunmuş / Task 1.2: VIP filled with False, existing values preserved"), (30, bool(ok_drop), "Görev 2: yalnızca PassengerId çıkarılmış / Task 2: only PassengerId removed")]

def check_hw3(df):
    return _show_results(_evaluate_hw3(df, globals()))

def _show_results(results):
    score = 0
    print("--- Ödev Test Sonuçları / Homework Test Results ---")
    for points, ok, label in results:
        print(("✅ Başarılı / Passed: " if ok else "❌ Hatalı veya eksik / Incorrect or missing: ") + label)
        if ok: score += points
    print(f"Toplam Puan / Total Score: {score}/100")
    return score
```
